```python
import pandas as pd

url = "http://prod.publicdata.landregistry.gov.uk.s3-website-eu-west-1.amazonaws.com/pp-2023.csv"

df = pd.read_csv(url, header=None)

df.head()
```

```python
df.columns = [
    "transaction_id",
    "price",
    "date",
    "postcode",
    "property_type",
    "new_build",
    "tenure",
    "paon",
    "saon",
    "street",
    "locality",
    "town_city",
    "district",
    "county",
    "ppd_category",
    "record_status"
]

df.head()
```

```python
df.info()
```

```python
df["date"] = pd.to_datetime(df["date"])

df.info()
```

```python
average_price = df["price"].mean()

print(f"Average UK house price: £{average_price:,.0f}")
```

```python
import matplotlib.pyplot as plt
```

```python
county_prices = (
    df.groupby("county")["price"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

county_prices
```

```python
county_prices.plot(kind="bar", figsize=(12,6))

plt.title("Top 10 Most Expensive Counties in UK")
plt.xlabel("County")
plt.ylabel("Average Price (£)")
plt.xticks(rotation=45)

plt.show()
```

```python
county_prices.plot(kind="bar", figsize=(12,6))

plt.title("Top 10 Most Expensive Counties in UK")
plt.xlabel("County")
plt.ylabel("Average Price (£)")
plt.xticks(rotation=45)

plt.savefig("top_10_counties.png", bbox_inches="tight")

plt.show()
```

```python
df["year"] = df["date"].dt.year

df[["date", "year"]].head()
```

```python
yearly_prices = (
    df.groupby("year")["price"]
    .mean()
)

yearly_prices
```

```python
yearly_prices.plot(figsize=(12,6))

plt.title("Average UK House Prices Over Time")
plt.xlabel("Year")
plt.ylabel("Average Price (£)")

plt.show()
```

```python
yearly_prices.plot(figsize=(12,6))

plt.title("Average UK House Prices Over Time")
plt.xlabel("Year")
plt.ylabel("Average Price (£)")

plt.savefig("uk_price_trends.png", bbox_inches="tight")

plt.show()
```

Insight

The analysis shows a clear upward trend in UK housing prices over time. Property values increased significantly after 2020, likely influenced by changes in housing demand, low interest rates, and post-pandemic market conditions.

```python
property_mapping = {
    "D": "Detached",
    "S": "Semi-Detached",
    "T": "Terraced",
    "F": "Flat",
    "O": "Other"
}

df["property_name"] = df["property_type"].map(property_mapping)

df[["property_type", "property_name"]].head()
```

```python
property_prices = (
    df.groupby("property_name")["price"]
    .mean()
    .sort_values(ascending=False)
)

property_prices
```

```python
property_prices.plot(kind="bar", figsize=(10,6))

plt.title("Average UK House Prices by Property Type")
plt.xlabel("Property Type")
plt.ylabel("Average Price (£)")
plt.xticks(rotation=0)

plt.show()
```

```python
property_prices.plot(kind="bar", figsize=(10,6))

plt.title("Average UK House Prices by Property Type")
plt.xlabel("Property Type")
plt.ylabel("Average Price (£)")
plt.xticks(rotation=0)

plt.savefig("property_type_prices.png", bbox_inches="tight")

plt.show()
```

Property Type Insight

Detached properties had the highest average prices across the UK housing market, while flats were generally the least expensive. This reflects differences in property size, land ownership, and location demand.

```python
london_data = df[df["county"].str.contains("LONDON", case=False, na=False)]

rest_uk_data = df[~df["county"].str.contains("LONDON", case=False, na=False)]

print("London Average Price: £{:,.0f}".format(london_data["price"].mean()))
print("Rest of UK Average Price: £{:,.0f}".format(rest_uk_data["price"].mean()))
```

```python
comparison = {
    "London": london_data["price"].mean(),
    "Rest of UK": rest_uk_data["price"].mean()
}

comparison_series = pd.Series(comparison)

comparison_series.plot(kind="bar", figsize=(8,6))

plt.title("London vs Rest of UK House Prices")
plt.ylabel("Average Price (£)")
plt.xticks(rotation=0)

plt.show()
```

```python
comparison_series.plot(kind="bar", figsize=(8,6))

plt.title("London vs Rest of UK House Prices")
plt.ylabel("Average Price (£)")
plt.xticks(rotation=0)

plt.savefig("london_vs_rest_uk.png", bbox_inches="tight")

plt.show()
```

Regional Insight

London property prices remain significantly higher than the rest of the UK housing market. This reflects strong demand, economic concentration, and limited housing supply within the capital compared to other regions.

```python
duplicates = df.duplicated().sum()

print(f"Duplicate rows: {duplicates}")
```

```python
df = df.dropna(subset=["price"])

df["price"].isnull().sum()
```

```python
df = df[df["price"] > 1000]

df["price"].describe()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.to_csv("cleaned_uk_housing.csv", index=False)

print("Cleaned dataset saved successfully.")
```

Data Cleaning Process

The dataset was cleaned by removing missing values, checking duplicate transactions, and filtering out unrealistic property prices below £1,000. Dates were converted into datetime format to support time-series analysis.

```python
df["month"] = df["date"].dt.month

df["is_new_build"] = df["new_build"].map({
    "Y": 1,
    "N": 0
})

df[["price", "year", "month", "is_new_build"]].head()
```

```python
correlation = df[[
    "price",
    "year",
    "month",
    "is_new_build"
]].corr()

correlation
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.savefig("correlation_heatmap.png", bbox_inches="tight")

plt.show()
```

Correlation Analysis

The heatmap shows relationships between housing market variables. Property prices displayed a moderate positive relationship with time, indicating long-term market growth. New-build properties showed slight differences in pricing compared to existing homes.
